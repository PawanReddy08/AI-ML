```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2005,2006,2007,2008])
y=np.array([25,34,54,12])
plt.plot(x,y,marker="^",
         markersize=10,
         markerfacecolor="red",
         markeredgecolor="cyan",
         linestyle="dashdot",
         linewidth="4",
         color="green")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x=np.array([2005,2006,2007,2008])
y=np.array([25,34,24,12])
z=np.array([23,20,43,30])
plt.title("class size",fontsize=15,
                       family="Arial",
                       fontweight="bold",
                       color="red")
plt.xlabel("years",fontsize=15,
                   family="arial",
                   fontweight="bold",
                   color="green")
plt.ylabel("students",fontsize=15,
                   family="arial",
                   fontweight="bold",
                   color="green")
plt.tick_params(axis="both",
                colors="brown")
plt.plot(x,z,marker=".")
plt.plot(x,y,marker=".")
plt.xticks(x)
plt.show()
```

```python
#grid() helps make plots easier to read by referencing lines
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,4,5])
y=np.array([5,10,15,20,25])
plt.plot(x,y)
plt.grid(axis="both",
         linewidth=3,
         linestyle="dotted",
         color="green")
plt.xticks(x)
plt.show()
```

```python
#BAR CHART  compare the categories of data by representing each category with a bar
import matplotlib.pyplot as plt
import numpy as np
categories=np.array(["fruits","vegetables","protein","dairy","sweets"])
values=np.array([3,4,5,3,1])
plt.bar(categories,values,color="red")
# plt.barh(categories,values) to represent in horizontal view
plt.title("daily consumption",fontsize=25,
                              family="arial",
                              fontweight="bold",
                              color="skyblue")
plt.xlabel("category",fontsize=15,
                      color="green",)
plt.ylabel("quantity",fontsize=15,
                      color="green")

```

```python
import matplotlib.pyplot as plt
import numpy as np
categories=["Freshmen","sophomores","seniors","juniors"]
values=np.array([300,275,325,255])
plt.pie(values,labels=categories,
               autopct="%1.1f%%",
               explode=[0,0,0,0.1],
               shadow=True,
               startangle=90)
plt.show()
```

```python
#SCATTERPLOT   shows the relationship bw two variables to identify a correlation(+,-,None)
# ex  study hours or test scores
import matplotlib.pyplot as plt
import numpy as np
a=np.array([0,1,2,3,3,4,5,5,6,7,8])
b=np.array([36,42,51,54,65,68,71,74,85,88,90])
x=np.array([0,1,2,2,3,4,5,5,5,7,7])
y=np.array([36,48,51,52,65,68,69,70,85,81,97])
plt.scatter(a,b,color="red",label="class a",alpha=0.5,s=100)
plt.scatter(x,y,color="blue",label="class b",alpha=0.5,s=100)
plt.legend()
# plt.grid()
plt.show()
```

```python
# HISTOGRAM  visual representation of the distribution of quantitative data. they group values into bins(no.of lines) and to calculate frequency.
import matplotlib.pyplot as plt
import numpy as np
scores=np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores,color="green",edgecolor="black",bins=25)
plt.show()
```

```python
# SUBPLOTS
import matplotlib.pyplot as plt
import numpy as np
x=np.array([1,2,3,4,5])
figure,axes=plt.subplots(2,2)
axes[0,0].bar(x,x*2,color="red")
axes[0,0].set_title("x*2")
axes[0,1].barh(x,x**2,color="green")
axes[0,1].set_title("x**2")
axes[1,0].hist(x,x**3,color="blue")
axes[1,0].set_title("x**3")
axes[1,1].scatter(x,x**4,color="purple")
axes[1,1].set_title("x**4")
plt.tight_layout()
plt.show()
```
