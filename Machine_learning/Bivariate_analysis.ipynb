```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips=sns.load_dataset('tips')
tips.head()
```

```python
titanic=pd.read_csv('train_titanic.csv')
titanic.head()
```

```python
flights=sns.load_dataset('flights')
flights.head()
```

```python
iris=sns.load_dataset('iris')
iris.head()
```

## Scatterplot (numerical-numerical)

```python
sns.scatterplot(x=tips['total_bill'],y=tips['tip'],hue=tips['sex'],style=tips['smoker'],size=tips['size'])
```

## Bar plot (Numerical-Categorical)

```python
sns.barplot(x=titanic['Pclass'],y=titanic['Age'],hue=titanic['Sex'])
```

## Boxplot (Numerical-Categorical))

```python
sns.boxplot(x=titanic['Sex'],y=titanic['Age'],hue=titanic['Survived'])
```

## Distplot (Numerical - Categorical)

```python
sns.displot(titanic[titanic['Survived']==0]['Age'],kde=True)
sns.displot(titanic[titanic['Survived']==1]['Age'],kde=True)
```

## Heatmap (Categorical-Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
sns.clustermap(pd.crosstab(titanic['SibSp'],titanic['Survived']))
```

## Pairplot

```python
sns.pairplot(tips,hue="sex")
```

## lineplot

```python
new=flights.groupby('year').sum('passengers').reset_index()
```

```python
sns.lineplot(x=new['year'],y=new['passengers'])
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
